```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [5, 8, 2, 6]

plt.figure(figsize=(4, 3)) # Adjust the figure size
plt.bar(x, y, color='red')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Simple Bar Plot')
plt.show()
```

```python
# Solve the system of equations:
# x + y = 10
# 2x - y = 5

# From the first equation, express y in terms of x:
# y = 10 - x

# Substitute this expression for y into the second equation:
# 2x - (10 - x) = 5
# 2x - 10 + x = 5
# 3x - 10 = 5
# 3x = 15
# x = 15 / 3
x = 5

# Now substitute the value of x back into the first equation to find y:
# y = 10 - x
y = 10 - x

print(f"The solution is x = {x} and y = {y}")
```
