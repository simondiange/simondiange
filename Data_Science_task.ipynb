**TASK 1 TECHCRUSH DATA SCIENCE** QUESTION 1

```python
 # 1) Create a list of 5 items you plan to buy.
shopping_list = ["Rice", "Beans", "Yam", "Plantain", "Garri"]

# 2) Create a dictionary where each item is the key, and its price (in Naira) is the value.
item_prices = {
    "Rice": 500,
    "Beans": 300,
    "Yam": 700,
    "Plantain": 400,
    "Garri": 250
}

# 3) Print the first and last item from your shopping list.
print("First item:", shopping_list[0])
print("Last item:", shopping_list[-1])

# 4) Print the price of the most expensive item from the dictionary.
most_expensive_item = max(item_prices, key=item_prices.get)
print("Price of the most expensive item:", item_prices[most_expensive_item])

# 5) Add 2 more items to the shopping list.
shopping_list.append("Tomatoes")
shopping_list.append("Onions")

# 6) Find the total cost of all the items.
total_cost = sum(item_prices.values())
print("Total cost of all items:", total_cost)

# 7) Create a tuple with your 3 favorite items from the shopping list and print it.
favorite_items = tuple(shopping_list[:3]) # Taking the first 3 items as favorites for this example
print("My 3 favorite items:", favorite_items)

# 8) Display the items in your shopping list in alphabetical order.
shopping_list.sort()
print("Shopping list in alphabetical order:", shopping_list)
```

**TASK 1 TECHCRUSH DATA SCIENCE** QUESTION 2

```python
# 1) Create variables to store student details
student_name = "Ngilla Simon Diange"
student_age = 29
student_gpa = 4.2
student_passed = True
```

```python
# 2) Print a sentence that includes all of the student’s details.
print(f"{student_name} is {student_age} years old, has a GPA of {student_gpa}, and has passed: {student_passed}.")
```

```python
# 3) Create a list of 5 subjects the student is taking. Print the 2nd and 4th subjects.
student_subjects = ["Math", "DataScience", "History", "English", "French"]
print("Second subject:", student_subjects[1])
print("Fourth subject:", student_subjects[3])
```

```python
# 4) Create a tuple of the student’s hobbies. Print the last hobby.
student_hobbies = ("Reading", "Coding", "Painting", "Gaming", "Football")
print("Last hobby:", student_hobbies[-1])
```

```python
# 5) Create a dictionary with the student’s details. Print the dictionary.
student_details_dict = {
    "name": student_name,
    "age": student_age,
    "gpa": student_gpa,
    "passed": student_passed,
    "subjects": student_subjects
}
print("Student details dictionary:", student_details_dict)
```

```python
# 6) Suppose you have another list for the test scores of the students.
# The scores are; 78, 80, 92, 56, 85. Find the average test score.
test_scores = [78, 80, 92, 56, 85]
average_score = sum(test_scores) / len(test_scores)
print("Average test score:", average_score)
```
