```python
from PIL import Image

try:
  img = Image.open('/content/+237 6 52 57 61 34 20250506_154218.jpg')
  print("Image loaded successfully!")
  display(img)
finally:
  pass
```

```python
print("Image Format:", img.format)
print("Image Size:", img.size)
print("Image Mode:", img.mode)
```

```python
# Check if the image is in RGB mode
if img.mode == 'RGB':
    # Split the image into its color bands
    r, g, b = img.split()

    print("Red Channel:")
    display(r)

    print("Green Channel:")
    display(g)

    print("Blue Channel:")
    display(b)
elif img.mode == 'RGBA':
    # Split the image into its color bands
    r, g, b, a = img.split()

    print("Red Channel:")
    display(r)

    print("Green Channel:")
    display(g)

    print("Blue Channel:")
    display(b)

    print("Alpha Channel:")
    display(a)
else:
    print(f"Image is in {img.mode} mode. Splitting into color channels is not directly supported for this mode.")
```

```python
from PIL import Image

# Define the rotation angle in degrees
angle_deg = 260

# Rotate the image
rotated_img = img.rotate(angle_deg, resample=Image.BICUBIC, expand=True)

# Display the rotated image
print(f"Image rotated by {angle_deg} degrees:")
display(rotated_img)
```

```python
from PIL import Image

# Define the rotation angle in degrees
angle_deg = 180

# Rotate the image
rotated_img_180 = img.rotate(angle_deg, resample=Image.BICUBIC, expand=True)

# Display the rotated image
print(f"Image rotated by {angle_deg} degrees:")
display(rotated_img_180)
```

```python
from PIL import Image

# Convert the image to grayscale ('L' mode for luminance)
gray_img = img.convert('L')

# Display the grayscale image
print("Grayscale Image:")
display(gray_img)
```
