```python
orders_df = pd.read_csv('pizza_sales/orders.csv')
order_details_df = pd.read_csv('pizza_sales/order_details.csv')
pizzas_df = pd.read_csv('pizza_sales/pizzas.csv')
pizza_types_df = pd.read_csv('pizza_sales/pizza_types.csv', encoding='latin-1')

print("orders_df shape:", orders_df.shape)
print("order_details_df shape:", order_details_df.shape)
print("pizzas_df shape:", pizzas_df.shape)
print("pizza_types_df shape:", pizza_types_df.shape)
```

## Join the data


Merge the four DataFrames into a single DataFrame based on the common columns.



Merge the four dataframes based on the common columns as instructed.



```python
merged_details_pizzas_df = pd.merge(order_details_df, pizzas_df, on='pizza_id')
merged_all_df = pd.merge(merged_details_pizzas_df, pizza_types_df, on='pizza_type_id')
df = pd.merge(merged_all_df, orders_df, on='order_id')
display(df.head())
df.info()
```

## Data cleaning and preparation


Inspect the joined DataFrame for any inconsistencies, missing values, or incorrect data types and perform necessary cleaning and transformations.



Check for missing values and data types in the merged DataFrame to identify areas needing cleaning or transformation.



```python
print("Missing values per column:")
print(df.isnull().sum())
print("\nData types of each column:")
df.info()
```


Based on the data types shown by `df.info()`, the 'date' and 'time' columns are currently objects (strings). To enable time-based analysis, they need to be converted to datetime objects. No missing values were found, so no imputation or dropping is necessary for missing data. All other data types seem appropriate for the columns they represent.



```python
df['date'] = pd.to_datetime(df['date'])
df['time'] = pd.to_datetime(df['time']).dt.time
display(df.head())
df.info()
```


The previous step successfully converted the 'date' column to datetime objects. However, the 'time' column was converted to `datetime.time` objects within a pandas `object` dtype column, as indicated by `df.info()`. While this is a valid representation of time, for potential time-series analysis or plotting of peak hours, it's often more convenient to have a full datetime object or extract relevant time components (like hour of the day). Given the goal of analyzing peak sales hours, converting the 'time' column to datetime objects (by combining with the date, though only the time part will be used for hourly analysis) and then extracting the hour will be most useful. The UserWarning also suggests explicitly specifying a format for more robust parsing.



```python
# Re-convert time with a specified format and then extract the time part
df['time'] = pd.to_datetime(df['time'], format='%H:%M:%S').dt.time

# To extract the hour for analysis, we can combine date and time into a single datetime column temporarily
df['datetime'] = pd.to_datetime(df['date'].astype(str) + ' ' + df['time'].astype(str))
df['hour'] = df['datetime'].dt.hour

display(df.head())
df.info()
```

## Perform sales analysis

Calculate total revenue/sales.



Calculate the total revenue by multiplying the 'quantity' and 'price' columns and then summing the result, and print the result.



```python
total_revenue = (df['quantity'] * df['price']).sum()
print(f"Total Revenue/Sales: ${total_revenue:,.2f}")
```

## Perform sales analysis


Find the total quantity sold.



Calculate the sum of the quantity column in the df DataFrame to find the total quantity of pizzas sold and then print the result.



```python
total_quantity_sold = df['quantity'].sum()
print(f"Total Quantity Sold: {total_quantity_sold}")
```

## Perform sales analysis


Find the total orders.



Count the number of unique order IDs to find the total number of orders and print the result.



```python
total_orders = df['order_id'].nunique()
print(f"Total Orders: {total_orders}")
```

## Perform sales analysis

Count the number of pizza types.



Calculate the number of unique values in the `pizza_type_id` column and print the result.



```python
num_pizza_types = df['pizza_type_id'].nunique()
print(f"Number of unique pizza types: {num_pizza_types}")
```

## Perform sales analysis


Find the average price of the pizzas.



Calculate the average price of the pizzas from the `price` column of the `df` DataFrame and print the result formatted to two decimal places.



```python
average_pizza_price = df['price'].mean()
print(f"Average Pizza Price: ${average_pizza_price:.2f}")
```

## Perform sales analysis


Determine the peak hours of sales.



Group the DataFrame by the 'hour' column and calculate the sum of 'quantity' for each hour, then sort the results to identify peak hours and print them.



```python
peak_hours = df.groupby('hour')['quantity'].sum().sort_values(ascending=False)
print("Peak Sales Hours:")
print(peak_hours)
```

## Perform sales analysis


Find total sales made on each day of the week and identify the busiest day.



Convert the 'date' column to datetime objects and extract the day of the week to find total sales per day.



```python
df['date'] = pd.to_datetime(df['date'])
df['day_of_week'] = df['date'].dt.day_name()
sales_by_day = df.groupby('day_of_week')['quantity'].sum().sort_values(ascending=False)
print("Total sales by day of the week:")
print(sales_by_day)
```

## Perform sales analysis


Find the top 5 bestselling pizzas.



Group the DataFrame by pizza_id, sum the quantity, sort in descending order, and select the top 5 to find the top 5 bestselling pizzas.



```python
top_5_bestselling_pizzas = df.groupby('pizza_id')['quantity'].sum().sort_values(ascending=False).head(5)
print("Top 5 Bestselling Pizzas:")
print(top_5_bestselling_pizzas)
```

## Perform sales analysis


Find the sales made in each month and identify any trends.



Convert the 'date' column to datetime objects, extract the month, group by month, sum the quantity, sort by month, and print the results to find the sales made in each month and identify trends.



```python
df['date'] = pd.to_datetime(df['date'])
df['month'] = df['date'].dt.month
sales_by_month = df.groupby('month')['quantity'].sum().sort_index()
print("Total sales by month:")
print(sales_by_month)
```

## Perform sales analysis


Identify pizza types that are not performing well.



Group the data by pizza type and sum the quantities, then sort in ascending order to identify underperforming pizzas and print the result.



```python
underperforming_pizzas = df.groupby('pizza_type_id')['quantity'].sum().sort_values(ascending=True)
print("Underperforming Pizza Types (by quantity sold):")
print(underperforming_pizzas)
```
