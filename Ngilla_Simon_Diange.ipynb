```python
# A) Create a list of student scores
student_scores = [85, 72, 90, 55, 40, 65, 78, 92, 48, 88]
```

```python
# B) Define the grade_student function
def grade_student(score):
    if score >= 80:
        return "A"
    elif 70 <= score < 80:
        return "B"
    elif 60 <= score < 70:
        return "C"
    elif 50 <= score < 60:
        return "D"
    else:
        return "F"
```

```python
# C) Use a loop to go through all student scores and call the function for each one.
# D) Print results in this format: "Student 1: Score = 85, Grade = A"
for i, score in enumerate(student_scores):
    grade = grade_student(score)
    print(f"Student {i + 1}: Score = {score}, Grade = {grade}")
```

```python
# A) Initialize the balance
balance = 1000
```

```python
# B) AND C) Define the atm_machine function
def atm_machine():
    global balance # Declare balance as global at the beginning
    # Implement the main loop
    while True:
        # i to iv Display options and get user input
        print("\nWelcome to the ATM!")
        print("Please choose an option:")
        print("1 → Check Balance")
        print("2 → Deposit Money")
        print("3 → Withdraw Money")
        print("4 → Exit")

        choice = input("Enter your choice: ")

        # D) I will now Implement conditional logic for actions
        if choice == '1':
            print(f"Your current balance is: naira{balance}")
        elif choice == '2':
            try:
                deposit_amount = float(input("Enter amount to deposit: "))
                if deposit_amount > 0:
                    balance += deposit_amount
                    print(f"Successfully deposited naira{deposit_amount}. Your new balance is: naira{balance}")
                else:
                    print("Invalid deposit amount.")
            except ValueError:
                print("Invalid input. Please enter a number.")
        elif choice == '3':
            try:
                withdraw_amount = float(input("Enter amount to withdraw: "))
                if withdraw_amount > 0:
                    if balance >= withdraw_amount:
                        balance -= withdraw_amount
                        print(f"Successfully withdrew naira{withdraw_amount}. Your new balance is: naira{balance}")
                    else:
                        print("Insufficient balance.")
                else:
                    print("Invalid withdrawal amount.")
            except ValueError:
                print("Invalid input. Please enter a number.")
        elif choice == '4':
            print("Thank you for using the ATM. Goodbye!")
            break
        else:
            print("Invalid choice. Please select a valid option.")

        # E) I will then call the atm_machine function (This will be done outside the function definition)
```
